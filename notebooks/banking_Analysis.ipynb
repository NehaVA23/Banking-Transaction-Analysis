```python
print("Calculate IQR")
Q1 = df['Transaction_Amount'].quantile(0.25)
Q3 = df['Transaction_Amount'].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
upper_bound = Q3 + 1.5 * IQR

print("Upper Outlier Bound:", upper_bound)
```

```python
account_activity = (
    df.groupby('Account_Number')
      .agg(
          Transaction_Count=('Transaction_ID', 'count'),
          Total_Amount=('Transaction_Amount', 'sum'),
          Average_Amount=('Transaction_Amount', 'mean')
      )
      .sort_values('Transaction_Count', ascending=False)
)

account_activity.head(10)
```

```python
top_accounts_by_amount = (
    account_activity
    .sort_values('Total_Amount', ascending=False)
)

top_accounts_by_amount.head(10)
```

```python
route_counts = (
    df.groupby(['Transaction_Source', 'Transaction_Destination'])
      .size()
      .reset_index(name='Transaction_Count')
      .sort_values('Transaction_Count', ascending=False)
)

route_counts.head(10)
```

```python
mode_type_analysis = (
    df.groupby(['Transaction_Mode', 'Transaction_Type'])
      .agg(
          Transaction_Count=('Transaction_ID', 'count'),
          Total_Amount=('Transaction_Amount', 'sum'),
          Average_Amount=('Transaction_Amount', 'mean')
      )
      .sort_values('Transaction_Count', ascending=False)
)

mode_type_analysis
```

```python
hourly_activity = (
    df.groupby('Hour')
      .agg(
          Transaction_Count=('Transaction_ID', 'count'),
          Total_Amount=('Transaction_Amount', 'sum'),
          Average_Amount=('Transaction_Amount', 'mean')
      )
      .sort_index()
)

hourly_activity
```

```python
day_analysis = (
    df.groupby('Day_Name')
      .agg(
          Transaction_Count=('Transaction_ID', 'count'),
          Total_Amount=('Transaction_Amount', 'sum'),
          Average_Amount=('Transaction_Amount', 'mean')
      )
)

day_analysis
```

```python
monthly_analysis = (
    df.groupby(['Year', 'Month', 'Month_Name'])
      .agg(
          Transaction_Count=('Transaction_ID', 'count'),
          Total_Amount=('Transaction_Amount', 'sum'),
          Average_Amount=('Transaction_Amount', 'mean')
      )
      .reset_index()
      .sort_values(['Year', 'Month'])
)

monthly_analysis
```

```python
mode_analysis = (
    df.groupby('Transaction_Mode')
      .agg(
          Transaction_Count=('Transaction_ID', 'count'),
          Total_Amount=('Transaction_Amount', 'sum'),
          Average_Amount=('Transaction_Amount', 'mean')
      )
      .sort_values('Transaction_Count', ascending=False)
)

mode_analysis
```

```python
type_analysis = (
    df.groupby('Transaction_Type')
      .agg(
          Transaction_Count=('Transaction_ID', 'count'),
          Total_Amount=('Transaction_Amount', 'sum'),
          Average_Amount=('Transaction_Amount', 'mean')
      )
      .sort_values('Transaction_Count', ascending=False)
)

type_analysis
```

```python
high_value_profile = (
    high_value_transactions
    .groupby(['Transaction_Mode', 'Transaction_Type'])
    .agg(
        Transaction_Count=('Transaction_ID', 'count'),
        Total_Amount=('Transaction_Amount', 'sum'),
        Average_Amount=('Transaction_Amount', 'mean')
    )
    .sort_values('Transaction_Count', ascending=False)
)

high_value_profile
```

```python
account_stats = (
    df.groupby('Account_Number')
      .agg(
          Transaction_Count=('Transaction_ID', 'count'),
          Total_Amount=('Transaction_Amount', 'sum'),
          Average_Amount=('Transaction_Amount', 'mean')
      )
      .reset_index()
)

account_stats.head()
```

```python
account_amount_p99 = account_stats['Total_Amount'].quantile(0.99)

print("Account Total Amount P99:", account_amount_p99)
```

```python
high_value_accounts = account_stats[
    account_stats['Total_Amount'] > account_amount_p99
].copy()

print("High-Value Accounts:", len(high_value_accounts))
```

```python
df['High_Value_Flag'] = (
    df['Transaction_Amount'] > p99
).astype(int)
```

```python
df['Risk_Score'] = df['High_Value_Flag']

df['Risk_Level'] = df['Risk_Score'].map({
    0: 'Low',
    1: 'High'
})
```

```python
risk_summary = (
    df['Risk_Level']
      .value_counts()
      .reset_index()
)

risk_summary.columns = ['Risk_Level', 'Transaction_Count']

risk_summary
```

```python
# Save the final cleaned dataset for MySQL and Power BI
output_path = "E:/Banking-Transaction-Analysis/data/cleaned_banking_transactions.csv"

df.to_csv(output_path, index=False)

print("Final cleaned dataset saved successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("File:", output_path)
```
